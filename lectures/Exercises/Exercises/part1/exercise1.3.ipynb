# A)
Use the binomial distribution ${{n}\choose{k}} p^k(1-p)^{n-k}$ to compute the probability of $k$ successes over $n$ observations of a binary random variable with probability $p$, for different values of $k,n,p$. Compute also the probability of a number of successes $\leq k$ (cumulative).

```python
import numpy as np
from scipy.stats import binom
import itertools

k = 2  # Number of successes
n = 4   # Number of trials
p = 0.5  # Probability of success
prob = binom.pmf(k, n, p) # Probability mass function
cum = binom.cdf(k, n, p)  # Cumulative distribution function
print(f'probability of {k} successes with probability {p} over {n} observations {prob}')
print(f'probability of 0 to {k} successes with probability {p} over {n} observations {cum}')
```

```python
obs_array = np.arange(n)  # the n-dimensional array [0,1,2,...,n-1]
combinations_object = itertools.combinations(obs_array, k)  
# All subsets of k elements of obs_array
combinations_list = list(combinations_object)  
# List of subsets of k elements [(0,1),(0,2),(1,2)]
n_comb = len(combinations_list)
print(f'number of combinations {n_comb}')
print(f'the combinations are {combinations_list}')
```

```python
successes_matrix = np.zeros((n_comb, n))
probability_matrix = (1-p)*np.ones((n_comb, n))
print(f'initial successes matrix \n{successes_matrix}')
print(f'\n initial probability matrix \n {probability_matrix}')

```

```python
for i in range(n_comb):
    successes_matrix[i, combinations_list[i]] = 1
    probability_matrix[i, combinations_list[i]] = p
    
print(f'successes matrix \n {successes_matrix}')
print(f'\n probability matrix \n {probability_matrix}')
```

```python
prob_comb_array = np.prod(probability_matrix, axis=1)  # Independent events
total_probability = np.sum(prob_comb_array)
print(f'probability of each combination \n {prob_comb_array}')
print(f'probability of {k} successess with probability {p} over {n} observations {total_probability}')
```

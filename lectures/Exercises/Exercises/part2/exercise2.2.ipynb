Use Pandas to import dataset 'Y.csv'. Plot pairwise scatter plots of some columns and compare the Pearson's coefficient.

```python
import numpy as np
import pandas as pd
from bokeh.plotting import figure, output_file, show
from bokeh.io import output_notebook


Y = pd.read_csv('./data_ex2.1-2.2/Y.csv', delimiter=',')
Y
```

```python
scattervar = list(Y.columns)
output_notebook()
TOOLS="hover,crosshair,pan,wheel_zoom,zoom_in,\
    zoom_out,box_zoom,undo,redo,reset,tap,save,\
    box_select,poly_select,lasso_select,"
var1 = scattervar[0]
for i in range(len(scattervar)-1):
        vari = scattervar[i+1]
        p = figure(tools=TOOLS, width=250, height=250, title=var1 + ' vs ' + vari)
        p.scatter(x=Y[var1], y=Y[vari])
        show(p)
        input('Press enter to continue...')
        print(type(np.corrcoef(Y[var1], Y[vari])[0, 1]))
        print(f'correlation of {var1} and {vari}:', np.corrcoef(Y[var1], Y[vari])[0, 1])
        input('Press enter to continue...')
        
var1 = scattervar[-2]
vari = scattervar[-1]
p = figure(tools=TOOLS, width=250, height=250, title=var1 + ' vs ' + vari)
p.scatter(x=Y[var1
          ], y=Y[vari])
show(p)
input('Press enter to continue...')
print(f'correlation of {var1} and {vari}:', np.corrcoef(Y[var1], Y[vari])[0, 1])
input('Press enter to continue...')

```

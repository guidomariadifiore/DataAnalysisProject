# A)
Load from sklearn.datasets the dataset of 6 handwritten digits. Apply the PCA to reduce the dimensionality to 2 and scatter plot the transformed data. Observe the Proportion of Explained Variance (PVE) for the Principal Components. Observe the loadings.

```python
# conda install scikit-learn
from IPython.core.display import display, HTML
display(HTML("<style>.container {​​​​​ width:200% !important; }​​​​​</style>"))
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA


digits = load_digits(n_class=6)
print(digits)
#print(digits.images, digits.images.shape)
#print(digits.data, digits.data.shape)
#print(digits.target, digits.target.shape)
```

```python
print(digits.images[0])

fig, ax = plt.subplots(4, 4, figsize=(6, 6))
for i, axi in enumerate(ax.flat): 
    axi.imshow(digits.images[i], cmap='binary') 
```

```python
pca_model = PCA(n_components=2)
pca_coordinates = pca_model.fit_transform(digits.data)
plt.scatter(pca_coordinates[:, 0], pca_coordinates[:, 1], c=digits.target, cmap=plt.cm.get_cmap('inferno',6))
plt.colorbar(ticks=range(6), label='digit value')
PVE = pca_model.explained_variance_ratio_
PVE_cum = np.sum(PVE)
print(f'PVE:{PVE}, cumulative PVE:{PVE_cum}')
```

```python
print(pca_model.components_)
```

# B)
Reduce the dimensionality to 10 and exploit the PVE to construct a scree plot. What do you observe?

```python
pca_model = PCA(n_components=10)
pca_coordinates = pca_model.fit_transform(digits.data)
plt.plot(pca_model.explained_variance_ratio_, linestyle='--', marker='o', color='b')
plt.title('scree plot')
plt.xlabel('ordered PCs')
plt.ylabel('PVE (%)')
print(pca_model.explained_variance_ratio_)
print(np.sum(pca_model.explained_variance_ratio_[:1]))
```

$\texttt{scikit-learn}$ provides a simple interface to build/use many Machine Learning models, with homogenous operations/structures regardless of the type of model. <br>
$\texttt{scikit-learn}$ can be used by applying the following steps
1. select the type of model (estimator) and import the corresponding class
2. initialize a model object of the chosen model class by setting the hyperparameters (configurations and model parameters not subject to training)
3. organize the training set in a feature matrix $\texttt{X}$ made up of $\texttt{n}$\_$\texttt{samples}$ rows and  $\texttt{n}$\_$\texttt{features}$ columns and target array (only for supervised learning) $\texttt{y}$ made up of $\texttt{n}$\_$\texttt{samples}$ rows
![alt text](Xy.png)
4. train the model on the training set with the $\texttt{fit()}$ method of the model object
5. generate predictions on new data with the method of the model object <br>
     a) $\texttt{predict()}$ (supervised and unsupervised learning) <br>
     b) $\texttt{transform()}$ (unsupervised learning)

# A)
Generate a dataset $X$ of $50$ random mono-dimensional points in the interval [0,10] and compute in these points the value of the function $y=2X-5+ \epsilon$ where $\epsilon$ is a random error. Perform the simple linear regression on the training set $(X,y)$.

```python
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(0)
X = 10 * np.random.rand(50) 
y = 2 * X - 5 + np.random.randn(50) 
plt.scatter(X, y);
X[:, np.newaxis].shape
print(X)
```

```python
from sklearn.linear_model import LinearRegression

linreg_model = LinearRegression(fit_intercept=True) 
linreg_model.fit(X[:, np.newaxis], y)

X_test = np.linspace(0,20, 1000) 
y_test_predicted = linreg_model.predict(X_test[:, np.newaxis])
plt.scatter (X, y)
plt.plot(X_test, y_test_predicted, c='r')
print(f'the estimated slope and intercept are {linreg_model.coef_} and {linreg_model.intercept_}')
```

# B)
Generate a dataset $X$ of $50$ random 2-dimensional points in the interval [0,10] and compute in these points the value of the function $y=2x_1-3x_2-5+ \epsilon$ where $\epsilon$ is a random error. Perform the multivariate linear regression on the training set $(X,y)$.

```python
X = 10 * np.random.rand(50, 2) 
y = 2 * X[:,0]- 3 * X[:,1] - 5 + np.random.randn(50)
multi_linreg_model = LinearRegression(fit_intercept=True) 
multi_linreg_model.fit(X, y) 
print(f'the estimated slopes and intercept are {multi_linreg_model.coef_} and {multi_linreg_model.intercept_}')
```

# C)
Generate a dataset $X$ of $50$ random mono-dimensional points in the interval [0,10] and compute in these points the value of the function $y=sin(x)+ 0.1\epsilon$ where $\epsilon$ is a random error. Use the $\texttt{PolynomialFeatures()}$ method of $\texttt{scikit-learn}$ to polynomially transform the dataset $X$ for a certain polynomial degree $q$. Apply the multivariate linear regression to the polynomially transformed points. Compare the $R^2$ score on the training set for different values of $q$.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

X = 10 * np.random.rand(50)
y = np.sin(X) + 0.1 * np.random.randn(50)
X=X[:, np.newaxis]
plt.scatter(X,y)
print(X)
```

```python
poly_object = PolynomialFeatures(degree=2)
X_poly = poly_object.fit_transform(X)
print(X_poly)
```

```python
linreg_model = LinearRegression(fit_intercept=True)
linreg_model.fit(X_poly, y)
y_predicted = linreg_model.predict(X_poly)

X_test = np.linspace(0, 10, 100)[:,np.newaxis]
X_test_poly = poly_object.fit_transform(X_test)
y_test_predicted = linreg_model.predict(X_test_poly)
plt.scatter(X, y, c='b')
plt.plot(X_test, y_test_predicted, c='r')

r2_train = r2_score(y,y_predicted)
print(r2_train)
```

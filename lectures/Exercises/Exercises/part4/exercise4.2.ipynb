# C)

Remove outliers by means of the $\sigma$-clipping technique, i.e.:

 - assume data is normally distributed
 
 - estimate $\sigma$ exploiting that for normal distributions $\hat\sigma=(q3-q1)/1.35$, and remove the tails (e.g. median-5$\sigma$,median+5$\sigma$)

Notice that the observed quartiles (q1, q3, and median(q2)) are used as, being ranking values (i.e., depending on the "position"), are less sensitive to ouliers, unlike $\sigma$ which is much more sensitive to anomlous values.

```python
a=np.random.randint(100,size=20)
print(a)
```

```python
print('mean:',np.mean(a))
print('sigma:',np.std(a))
print('median:',np.median(a)) 
```

```python
a[np.argmax(a)]=1000
print(a)
```

```python
print('mean:',np.mean(a))
print('sigma:',np.std(a))
print('median:',np.median(a)) 
```

```python
quartiles = np.percentile(df_births['births'], [25, 50, 75])
quartiles
```

```python
median=quartiles[1]
iqr=quartiles[2]-quartiles[0]
sigma_hat=iqr/1.35
sigma_hat
```

```python
df_births_clean=df_births.loc[(df_births['births']<median+5*sigma_hat) \
                          & (df_births['births']>median-5*sigma_hat),:]
df_births_clean.isnull().any()
```

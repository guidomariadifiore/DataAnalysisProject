# A)
Import the $\texttt{wine}$ dataset and use the methods $\texttt{DecisionTreeClassifier()}$  and $\texttt{plot\_tree()}$ of $\texttt{scikit-learn}$ to train, test and depict a 3 class classification tree.

```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

wine_dict = load_wine()
data = wine_dict['data']
target = wine_dict['target']
feat_names = wine_dict['feature_names']
targ_names = wine_dict['target_names']
print(data)
print(data.shape)
print(target)
print(feat_names)
```

```python
np.random.seed(11)
X_train, X_test, y_train, y_test = train_test_split(data, target, train_size=0.8) 
print(X_train.shape)
tree_model = DecisionTreeClassifier(criterion = 'entropy')
tree_model.fit(X_train, y_train)
y_predicted_train = tree_model.predict(X_train)
y_predicted_test = tree_model.predict(X_test)
training_acc = accuracy_score(y_train, y_predicted_train)
testing_acc = accuracy_score(y_test, y_predicted_test)
print(f'training accuracy {training_acc}\n testing accuracy {testing_acc}')
```

```python
fig, ax = plt.subplots(figsize=(20, 10))  
plot_tree(tree_model, ax=ax, fontsize=10, \
          feature_names=feat_names, class_names=targ_names, filled=True)
plt.show()
```

# B)
Apply the pruning by penalizing the tree complexity with the optional parameter $\texttt{ccp}$_$\texttt{alpha}$ according to the penalized function $\sum\limits_{p=1}^{|T|} \Big(-\sum\limits_{k=1,K}f_k \log(f_k) \Big)+ \alpha |T|$ and observe the difference with respect to the non-penalized one.

```python
tree_model = DecisionTreeClassifier(criterion = 'entropy', ccp_alpha=0.1)
tree_model.fit(X_train, y_train)
y_predicted_train = tree_model.predict(X_train)
y_predicted_test = tree_model.predict(X_test)
training_acc = accuracy_score(y_train, y_predicted_train)
testing_acc = accuracy_score(y_test, y_predicted_test)
print(f'training accuracy {training_acc}\n testing accuracy {testing_acc}')
fig, ax = plt.subplots(figsize=(20, 10))  
plot_tree(tree_model, ax=ax, fontsize=10, \
          feature_names=feat_names, class_names=targ_names, filled=True)
plt.show()
```

# C)
Create a 1000 samples 10 feature 5 class noisy dataset and train a classification tree without pruning and with pruning with grid-search cross-validation on the complexity parameter. Observe the difference in terms of bias and variance.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import GridSearchCV

np.random.seed(7)

X, y = make_classification(n_samples=1000, n_features=10, n_classes=5,\
                           n_clusters_per_class=1, flip_y=0.15, \
                           n_informative=8, n_redundant=2)
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8)

tree_model = DecisionTreeClassifier(criterion = 'entropy')
tree_model.fit(X_train, y_train)
y_predicted_train = tree_model.predict(X_train)
y_predicted_test = tree_model.predict(X_test)
training_acc = accuracy_score(y_train, y_predicted_train)
testing_acc = accuracy_score(y_test, y_predicted_test)
print(f'training accuracy {training_acc}\n testing accuracy {testing_acc}')
fig, ax = plt.subplots(figsize=(20, 10))  
plot_tree(tree_model, ax=ax, fontsize=10, filled=True)
plt.show()

```

```python
grid_of_values ={'ccp_alpha':[1, 0, 0.1, 0.01, 0.001]}
grid_model = GridSearchCV(estimator=DecisionTreeClassifier(criterion='entropy'), param_grid=grid_of_values, cv=5)
grid_model.fit(X_train,y_train)
best_alpha = grid_model.best_params_['ccp_alpha']
print(f'best alpha {best_alpha}')
tree_model = DecisionTreeClassifier(criterion = 'entropy', ccp_alpha=best_alpha)
tree_model.fit(X_train, y_train)

y_predicted_train = tree_model.predict(X_train)
y_predicted_test = tree_model.predict(X_test)
training_acc = accuracy_score(y_train, y_predicted_train)
testing_acc = accuracy_score(y_test, y_predicted_test)
print(f'training accuracy {training_acc}\n testing accuracy {testing_acc}')
fig, ax = plt.subplots(figsize=(20, 10))  
plot_tree(tree_model, ax=ax, fontsize=10, filled=True)
plt.show()
```
